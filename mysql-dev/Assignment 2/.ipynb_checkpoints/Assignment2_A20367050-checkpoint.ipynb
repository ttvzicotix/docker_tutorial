```python
query_create_order_item = """
CREATE TABLE IF NOT EXISTS OrderItem (
    Order_Num CHAR(8),
    Item_ID VARCHAR(255),
    Item_Qty INTEGER,

    PRIMARY KEY (Order_Num, Item_ID),

    FOREIGN KEY (Order_Num)
        REFERENCES CustomerOrder(Order_Num)
        ON UPDATE CASCADE
        ON DELETE CASCADE,

    FOREIGN KEY (Item_ID)
        REFERENCES Item(Item_ID)
        ON UPDATE CASCADE
        ON DELETE CASCADE
);
"""

cursor.execute(query_create_order_item)

print("OrderItem table created!")
```

```python
import csv
from pathlib import Path

data_folder = Path("Assignment 2/Data")
csv_files = sorted(data_folder.glob("*.csv"))
suffix = "_50"

for file in csv_files:
    print(file.name)

print("Total CSV files:", len(csv_files))
```

```python
def load_csv(filename, table_name, personalize_columns=()):

    with open(data_folder / filename,
              newline="", encoding="utf-8-sig") as file:

        reader = csv.DictReader(file)

        columns = reader.fieldnames

        column_names = ", ".join(f"`{col}`" for col in columns)
        placeholders = ", ".join(["%s"] * len(columns))

        query_insert = f"""
        INSERT INTO `{table_name}` ({column_names})
        VALUES ({placeholders})
        """

        records = []

        for row in reader:

            for col in personalize_columns:
                row[col] = row[col] + suffix

            values = tuple(
                row[col] if row[col] != "" else None
                for col in columns
            )

            records.append(values)

        cursor.executemany(query_insert, records)
        connection.commit()

        print(f"{table_name}: {len(records)} rows inserted")
```

```python
#Part 4
#Q1) Which items have the highest estimated annual sales revenue?

query_4_1 = """
SELECT
    Item_ID,
    Item_Description,
    Item_Price,
    Item_AnnualSales,
    Item_Price * Item_AnnualSales AS Estimated_Revenue
FROM Item
ORDER BY Estimated_Revenue DESC;
"""

cursor.execute(query_4_1)

for row in cursor.fetchall():
    print(row)
```

```python
#Q2) Which customers have placed the most orders?

query_4_2 = """
SELECT
    Customer.Customer_Num,
    Customer.Customer_FirstName,
    Customer.Customer_LastName,
    COUNT(CustomerOrder.Order_Num) AS Total_Orders
FROM Customer
LEFT JOIN CustomerOrder
    ON Customer.Customer_Num = CustomerOrder.Customer_Num
GROUP BY
    Customer.Customer_Num,
    Customer.Customer_FirstName,
    Customer.Customer_LastName
ORDER BY Total_Orders DESC;
"""

cursor.execute(query_4_2)

for row in cursor.fetchall():
    print(row)
```

```python
#Q3) How many units of each product were received from each vendor?

query_4_3 = """
SELECT
    Vendor.Vendor_Num,
    Vendor.Vendor_Name,
    Item.Item_ID,
    Item.Item_Description,
    SUM(ShipmentInfo.Item_Qty) AS Total_Units_Received
FROM Vendor
JOIN Shipment
    ON Vendor.Vendor_Num = Shipment.Vendor_Num
JOIN ShipmentInfo
    ON Shipment.Shipment_Num = ShipmentInfo.Shipment_Num
JOIN Item
    ON ShipmentInfo.Item_ID = Item.Item_ID
GROUP BY
    Vendor.Vendor_Num,
    Vendor.Vendor_Name,
    Item.Item_ID,
    Item.Item_Description
ORDER BY
    Vendor.Vendor_Num,
    Total_Units_Received DESC;
"""

cursor.execute(query_4_3)

for row in cursor.fetchall():
    print(row)
```

```python
#Q4) Which shipping carriers handled the most customer orders?

query_4_4 = """
SELECT
    Order_Carrier,
    COUNT(*) AS Total_Orders
FROM CustomerOrder
GROUP BY Order_Carrier
ORDER BY Total_Orders DESC;
"""

cursor.execute(query_4_4)

for row in cursor.fetchall():
    print(row)
```
